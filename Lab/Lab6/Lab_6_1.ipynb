# Lab 6.1: Simple Calculator Functions

## 1. Define the functions

```python
def add(a, b):
    """Return the sum of a and b."""
    return a + b


def subtract(a, b):
    """Return the result of a minus b."""
    return a - b


def multiply(a, b):
    """Return the product of a and b."""
    return a * b


def divide(a, b):
    """Return a divided by b, or an error message if b is zero."""
    if b == 0:
        return "Error: Division by zero"
    return a / b


def power(base, exponent=2):
    """Return base raised to exponent (default exponent is 2, i.e. squaring)."""
    return base ** exponent
```

## 2. Quick test of each function

```python
print(add(10, 5))
print(subtract(10, 5))
print(multiply(10, 5))
print(divide(10, 5))
print(divide(10, 0))   # division by zero handled
print(power(3))        # default exponent = 2
print(power(2, 3))     # exponent = 3
```

## 3. Menu-driven calculator (while loop)

```python
def show_menu():
    """Print the calculator menu."""
    print("\n===== Simple Calculator =====")
    print("1. Add")
    print("2. Subtract")
    print("3. Multiply")
    print("4. Divide")
    print("5. Power (exponent optional, default 2)")
    print("6. Exit")


while True:
    show_menu()
    choice = input("Choose an operation (1-6): ").strip()

    if choice == "6":
        print("Goodbye!")
        break
    if choice not in ("1", "2", "3", "4", "5"):
        print("Invalid choice. Please enter a number from 1 to 6.")
        continue

    try:
        if choice == "5":
            base = float(input("Enter base: "))
            exp_text = input("Enter exponent (press Enter for 2): ").strip()
            result = power(base) if exp_text == "" else power(base, float(exp_text))
        else:
            num1 = float(input("Enter first number: "))
            num2 = float(input("Enter second number: "))
            if choice == "1":
                result = add(num1, num2)
            elif choice == "2":
                result = subtract(num1, num2)
            elif choice == "3":
                result = multiply(num1, num2)
            else:
                result = divide(num1, num2)
    except ValueError:
        print("Invalid number. Please try again.")
        continue

    print(f"Result: {result}")
```
