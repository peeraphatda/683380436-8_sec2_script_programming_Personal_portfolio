# Activity 6.1: Refactoring a Previous Lab Assignment

Lab chosen: **Week 2 - Number Classifier**

Problem in the original code: the same if/else blocks are repeated (and copy-pasted again when classifying another number).

## Before refactoring

```python
num = float(input("Enter a number: "))

if num > 0:
    print("The number is positive.")
elif num < 0:
    print("The number is negative.")
else:
    print("The number is zero.")

if num % 2 == 0:
    print("The number is even.")
else:
    print("The number is odd.")

if num > 0:
    print("Absolute value:", num)
elif num < 0:
    print("Absolute value:", -num)
else:
    print("Absolute value:", 0)

another = input("Classify another number? (y/n): ")
if another == "y":
    num = float(input("Enter a number: "))
    if num > 0:
        print("The number is positive.")
    elif num < 0:
        print("The number is negative.")
    else:
        print("The number is zero.")
    if num % 2 == 0:
        print("The number is even.")
    else:
        print("The number is odd.")
```

## After refactoring (using functions)

```python
def is_positive_negative_zero(num):
    """Return 'positive', 'negative' or 'zero' for the given number."""
    if num > 0:
        return "positive"
    elif num < 0:
        return "negative"
    return "zero"


def is_even_odd(num):
    """Return 'even' or 'odd' for the given number."""
    return "even" if num % 2 == 0 else "odd"


def absolute_value(num):
    """Return the absolute value of the given number."""
    return -num if num < 0 else num


def read_number():
    """Prompt until the user enters a valid number, then return it as a float."""
    while True:
        try:
            return float(input("Enter a number: "))
        except ValueError:
            print("Invalid input. Please enter a number.")


def classify_number(num):
    """Print the full classification of a number."""
    print(f"The number is {is_positive_negative_zero(num)}.")
    print(f"The number is {is_even_odd(num)}.")
    print(f"Absolute value: {absolute_value(num)}")


classify_number(read_number())
while input("Classify another number? (y/n): ").strip().lower() == "y":
    classify_number(read_number())
print("Done!")
```

## Summary
- Repeated logic now lives in 3 small functions, each written once and reused.
- The code is easier to read, test and debug.
- Adding a feature (e.g. input validation) only needs a change in one place.
