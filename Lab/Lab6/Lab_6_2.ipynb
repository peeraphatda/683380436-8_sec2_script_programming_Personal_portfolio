# Lab 6.2: Custom Module and Standard Library Usage

## Part 1: Create a custom module (`my_utils.py`)

```python
%%writefile my_utils.py
# my_utils.py - custom module

def greet(name):
    """Print a greeting message for the given name."""
    print(f"Hello, {name}!")


def is_prime(number):
    """Return True if number is prime, otherwise False."""
    if number < 2:
        return False
    for i in range(2, int(number ** 0.5) + 1):
        if number % i == 0:
            return False
    return True
```

## Part 2: Use the custom module and standard modules

```python
import my_utils
import math
import random

my_utils.greet("Alice")

for n in (7, 10):
    print(f"is_prime({n}) = {my_utils.is_prime(n)}")

print(f"Square root of 144 = {math.sqrt(144)}")
print(f"Random number (1-100): {random.randint(1, 100)}")
```
