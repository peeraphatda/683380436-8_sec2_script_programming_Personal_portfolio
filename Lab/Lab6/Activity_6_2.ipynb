# Activity 6.2: Benefits of Modular Programming & Standard Library Exploration

## Part 1: Benefits of modularity
- **Large projects:** split into small pieces (functions/modules) so each part is easy to understand.
- **Teamwork:** each person works on a separate module without editing the same code.
- **Debugging:** an error can be traced to a specific function instead of searching the whole program.
- **Code testing:** each function can be tested on its own.
- **Reusability (DRY):** write once, use in many places or many projects.
- **Readability:** meaningful function names describe what the code does.

## Part 2: Standard library exploration
Docs: https://docs.python.org/3/library/

Modules explored: `datetime`, `os`, `collections`

### 1) `datetime` - work with dates and times

```python
from datetime import datetime, date

now = datetime.now()
print("Now:", now.strftime("%Y-%m-%d %H:%M:%S"))

birthday = date(2000, 1, 1)
age_days = (date.today() - birthday).days
print("Days since 2000-01-01:", age_days)
```

**Use case:** timestamps in logs, calculating age or deadlines.

### 2) `os` - interact with the operating system

```python
import os

print("Current folder:", os.getcwd())
print("Files here:", os.listdir("."))
print("Path exists?", os.path.exists("my_utils.py"))
```

**Use case:** list, check or manage files and folders automatically.

### 3) `collections` - special data structures

```python
from collections import Counter

words = ["apple", "banana", "apple", "cherry", "banana", "apple"]
count = Counter(words)
print(count)
print("Most common:", count.most_common(1))
```

**Use case:** quickly count how often each item appears (e.g. word frequency).
