```python
import math

point1 = (3, 4)
point2 = (6, 8)

print(f"Point 1: x = {point1[0]}, y = {point1[1]}")
print(f"Point 2: x = {point2[0]}, y = {point2[1]}")

print("\nNote: Attempting to change point1[0] = 5 results in a TypeError because tuples are immutable.")

distance = math.sqrt((point2[0] - point1[0])**2 + (point2[1] - point1[1])**2)
print(f"Calculated distance between point1 and point2: {distance}")

print("\n--- Challenge: Custom Points ---")
try:
    x1 = float(input("Enter x1: "))
    y1 = float(input("Enter y1: "))
    x2 = float(input("Enter x2: "))
    y2 = float(input("Enter y2: "))

    user_point1 = (x1, y1)
    user_point2 = (x2, y2)

    user_distance = math.sqrt((user_point2[0] - user_point1[0])**2 + (user_point2[1] - user_point1[1])**2)
    print(f"The distance between {user_point1} and {user_point2} is: {user_distance:.2f}")
except ValueError:
    print("Invalid input. Please enter numbers only.")
```
