# **Activity 4.1: List vs. Tuple: Scenario Matching**

1. A collection of daily temperatures recorded over a week. - อุณหภูมิรายวันในหนึ่งสัปดาห์

Ans. List เพราะ 1 สัปดาห์มี 7 วัน และไม่สามารถระบุอย่างแน่นอนได้ว่าอุณหภูมิจะคงที่เสมอจึงต้องใช้ list ที่สามารถ เปลี่ยนแปลงอย่างยืดหยุ่นได้

2. The RGB color values for a specific shade (e.g., (255, 0, 0) for red). - ค่าสี RGB ของเฉดสีเฉพาะ

Ans. Tuple เพราะ สามารถล็อกค่าของ r / g / b ไม่ให้แก้ไขโดยไม่ตั้งใจได้

3. Items in a shopping cart that can be added or removed. - สินค้าในรถเข็นช้อปปิ้ง

Ans. List เพราะ ลูกค้าแต่ละคนย่อมไม่หยิบของเหมือนกันทุกอย่างจึงต้องใช้งานตัวที่สามารถเปลี่ยนแปลง ลบ บวก ได้ตลอดเวลา

4. The days of the week in order. - วันในรอบสัปดาห์ตามลำดับ

Ans. Tuple เพราะ วัน จันทร์- อาทิตย์ เป็นอะไรที่ถูกระบุได้อย่างตายตัวจึงใช้ Tuple เนื่องจากไม่จำเป็นต้องเปลี่ยนแปลง

5. Student scores on a series of assignments, which might be updated. - คะแนนสอบของนักเรียน

Ans. List เพราะคะแนนของนักเรียนแต่ละคนล้วนไม่เท่ากันไม่ว่าจะเป็นคะแนนเสริมหรือทำงานได้ดีถูกตามเกณฑ์จึงใช้ list

6. GPS coordinates for a fixed landmark. - พิกัด GPS ของสถานที่สำคัญ

Ans. Tuple เพราะพิกัดของสถานที่นั้นๆเป็นตายตัวไม่เคลือนที่ไปไหน ดังพิกัดจึงตายตัวและใช้ Tuple
