```python
student_names = []

while True:
    print("\n--- Student Management System ---")
    print("1. Add Students")
    print("2. Find a Student")
    print("3. Remove a Student")
    print("4. Sort Students")
    print("5. Count Students")
    print("6. Exit")

    choice = input("Enter your choice (1-6): ").strip()

    if choice == '1':
        print("\n--- Add Students ---")
        for i in range(3):
            name = input(f"Enter name for student {i+1}: ").strip()
            student_names.append(name)
        print("Current list:", student_names)

    elif choice == '2':
        print("\n--- Find a Student ---")
        search_name = input("Enter the name to search for: ").strip()
        if search_name in student_names:
            idx = student_names.index(search_name)
            print(f"Found! {search_name} is at index {idx}.")
        else:
            print(f"{search_name} is not in the list.")

    elif choice == '3':
        print("\n--- Remove a Student ---")
        remove_name = input("Enter the name to remove: ").strip()
        if remove_name in student_names:
            student_names.remove(remove_name)
            print(f"Successfully removed {remove_name}.")
        else:
            print(f"Error: {remove_name} not found in the list.")
        print("Updated list:", student_names)

    elif choice == '4':
        print("\n--- Sort Students ---")
        student_names.sort()
        print("Sorted list (alphabetical):", student_names)

    elif choice == '5':
        print("\n--- Count Students ---")
        print(f"Total number of students: {len(student_names)}")

    elif choice == '6':
        print("Exiting program. Goodbye!")
        break
    else:
        print("Invalid choice. Please enter a number between 1 and 6.")
```
