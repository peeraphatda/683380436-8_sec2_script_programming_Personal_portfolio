```python
todo_list = []

while True:
    print("\n--- To-Do List Manager ---")
    print("1. Add a task")
    print("2. View tasks")
    print("3. Mark task as complete")
    print("4. Exit")

    choice = input("Choose an option (1-4): ").strip()

    if choice == '1':
        task = input("Enter the task description: ").strip()
        if task:
            todo_list.append(task)
            print(f"Added: '{task}'")
        else:
            print("Task cannot be empty.")

    elif choice == '2':
        print("\n--- Your Tasks ---")
        if len(todo_list) == 0:
            print("Your to-do list is empty!")
        else:
            for index, task in enumerate(todo_list, start=1):
                print(f"{index}. {task}")

    elif choice == '3':
        if len(todo_list) == 0:
            print("No tasks available to complete.")
            continue

        print("\nHow would you like to remove the completed task?")
        print("1. By Task Number (Index)")
        print("2. By Exact Task Name")
        remove_type = input("Choose 1 or 2: ").strip()

        if remove_type == '1':
            try:
                task_num = int(input("Enter the task number to complete: "))
                if 1 <= task_num <= len(todo_list):
                    removed = todo_list.pop(task_num - 1)
                    print(f"Completed and removed: '{removed}'")
                else:
                    print("Invalid task number.")
            except ValueError:
                print("Please enter a valid integer.")

        elif remove_type == '2':
            task_name = input("Enter the exact name of the task: ").strip()
            if task_name in todo_list:
                todo_list.remove(task_name)
                print(f"Completed and removed: '{task_name}'")
            else:
                print(f"Task '{task_name}' not found in your list.")
        else:
            print("Invalid option.")

    elif choice == '4':
        print("Goodbye! Stay productive.")
        break
    else:
        print("Invalid choice. Please enter a option between 1 and 4.")
```
