```python
number = int(input("Enter a number: "))
print(f"Multiplication table for {number}:")
for i in range(1, 13):
    result = number * i
    print(f"{number} x {i} = {result}")
```

```python
print("------- 12x12 Multiplication Grid -------")
for i in range(1, 13):
    for j in range(1, 13):
        print(f"{i * j:4d}", end="")
    print()
```
