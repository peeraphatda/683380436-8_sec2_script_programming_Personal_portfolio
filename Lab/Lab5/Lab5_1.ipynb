```python
def contact_book():
    contacts = {}

    while True:
        print("\n==============================")
        print("  Contact Book  ")
        print("==============================")
        print("1. Add/Update Contact)")
        print("2. View Contact Details")
        print("3. List All Contacts")
        print("4. Delete Contact")
        print("5. Exit")
        print("------------------------------")

        choice = input("เลือกเมนู (1-5): ").strip()


        if choice == '1':
            name = input("ระบุชื่อผู้ติดต่อ: ").strip().title()
            phone = input("ระบุเบอร์โทรศัพท์: ").strip()
            email = input("ระบุอีเมล: ").strip()

            contacts[name] = {
                "phone": phone,
                "email": email
            }
            print(f"บันทึกข้อมูลของ '{name}' เรียบร้อยแล้ว!")
        elif choice == '2':
            name = input("ระบุชื่อผู้ติดต่อที่ต้องการค้นหา: ").strip().title()
            details = contacts.get(name)

            if details:
                print(f"\n ข้อมูลผู้ติดต่อ: {name}")
                print(f"   • เบอร์โทรศัพท์: {details['phone']}")
                print(f"   • อีเมล: {details['email']}")
            else:
                print(" ไม่พบข้อมูลผู้ติดต่อนี้ในระบบ")

        elif choice == '3':
            if not contacts:
                print(" สมุดโทรศัพท์ยังไม่มีข้อมูล")
            else:
                print("\nรายชื่อผู้ติดต่อทั้งหมด:")
                for name, info in contacts.items():
                    print(f" • {name} -> เบอร์: {info['phone']}, อีเมล: {info['email']}")

        elif choice == '4':
            name = input("ระบุชื่อผู้ติดต่อที่ต้องการลบ: ").strip().title()

            if name in contacts:
                del contacts[name]
                print(f" ลบข้อมูลของ '{name}' ออกจากระบบเรียบร้อยแล้ว")
            else:
                print(" ไม่พบข้อมูลผู้ติดต่อนี้ในระบบ")

        elif choice == '5':
            print("ปิดการทำงานของโปรแกรม ")
            break

        else:
            print(" ตัวเลือกไม่ถูกต้อง กรุณากรอกหมายเลข 1-5 เท่านั้น")

if __name__ == "__main__":
    contact_book()
```
