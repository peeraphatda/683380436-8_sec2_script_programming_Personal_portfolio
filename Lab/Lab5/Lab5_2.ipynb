```python

python_students = {"S001", "S003", "S005", "S007", "S009"}
java_students = {"S002", "S003", "S006", "S007", "S010"}

print("=== ข้อมูลตั้งต้น ===")
print(f"นักศึกษาที่เรียน Python : {python_students}")
print(f"นักศึกษาที่เรียน Java   : {java_students}\n")

both_courses = python_students.intersection(java_students)
print(f"1. นักศึกษาที่เรียนทั้งสองวิชา (Intersection): {both_courses}")

all_students = python_students.union(java_students)
print(f"2. นักศึกษาทั้งหมดรวมกัน : {all_students}")

only_python = python_students.difference(java_students)
print(f"3. นักศึกษาที่เรียนแค่ Python เท่านั้น : {only_python}")

exclusive_students = python_students.symmetric_difference(java_students)
print(f"4. นักศึกษาที่เรียนแค่วิชาเดียวเท่านั้น : {exclusive_students}\n")

all_students_list = ["S001", "S003", "S005", "S001", "S002"]
print("=== ลบข้อมูลซ้ำใน List ===")
print(f"List ดั้งเดิม (มีข้อมูลซ้ำ) : {all_students_list}")

unique_students_list = list(set(all_students_list))
print(f"List หลังลบข้อมูลซ้ำแล้ว    : {unique_students_list}")
```
